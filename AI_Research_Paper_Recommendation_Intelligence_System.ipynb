Novelty / Outlier Detection (embedding-space)

```python
normalized_embeddings = embeddings.copy().astype('float32')
faiss.normalize_L2(normalized_embeddings)
```

```python
k = 6
D, I = index.search(normalized_embeddings, k)
```

```python
neighbor_scores = D[:, 1:]   #drop column 0 (self-match)
```

```python
avg_similarity = neighbor_scores.mean(axis=1)
df['novelty_score'] = 1 - avg_similarity
```

```python
print("=== MOST NOVEL / OUTLIER PAPERS ===")
print(df.sort_values('novelty_score', ascending=False)[['title', 'novelty_score']].head(10))

print("\n=== MOST 'GENERIC' / OVERSATURATED-TOPIC PAPERS ===")
print(df.sort_values('novelty_score', ascending=True)[['title', 'novelty_score']].head(10))
```

Auto Topic Discovery

```python
from sklearn.cluster import KMeans
```

```python
N_CLUSTERS = 15
```

```python
kmeans = KMeans(
    n_clusters=N_CLUSTERS,
    random_state=42,
    n_init=10
)
```

```python
cluster_labels = kmeans.fit_predict(normalized_embeddings)
df['cluster'] = cluster_labels
```

```python
cluster_topics = {}
```

```python
for cluster_id in range(N_CLUSTERS):
    cluster_indices = np.where(cluster_labels == cluster_id)[0]

    centroid = kmeans.cluster_centers_[cluster_id].reshape(1, -1)

    cluster_embeddings = normalized_embeddings[cluster_indices]
    distances = np.linalg.norm(cluster_embeddings - centroid, axis=1)

    closest_local_idx = distances.argsort()[:5]
    closest_global_idx = cluster_indices[closest_local_idx]

    representative_text = " ".join(df.iloc[closest_global_idx]['title'].tolist())

    keywords = kw_model.extract_keywords(
        representative_text,
        keyphrase_ngram_range=(1, 3),
        stop_words='english',
        top_n=5
    )
    topic_label = ", ".join([kw[0] for kw in keywords])
    cluster_topics[cluster_id] = topic_label

    print(f"Cluster {cluster_id} ({len(cluster_indices)} papers): {topic_label}")
```

```python
df['cluster_topic'] = df['cluster'].map(cluster_topics)

print(df[['title', 'cluster_topic']].sample(10))
```
